**Gains Chart**

**A gains chart is used to visualize the effectiveness of a predictive model. Here, it compares the cumulative number of positive outcomes (in this case, defaulting customers) against the number of records (x-axis). The dashed diagonal line represents a random model, which captures a proportional number of positives as you move through the records, while the solid blue line represents the performance of our predictive model. This chart indicates that our model performs better than random, as evidenced by the blue curve being consistently above the baseline dashed line. However, the model performance. However, the bigger the gap between the dashed lines and the blue line, the higher the performance, which is not so much the case with our model.**

**Lift Chart**

**The decile lift chart measures how better the model is at predicting positive outcomes than random guessing. The model captures 2.1 times more positive outcomes in the top 10% of records than would be expected by random chance. This indicates high effectiveness at this level.**

**When predicting loan default rates, recall should generally be prioritized over precision. Here's why:**
**- Prioritize Recall: Ensuring that the model has a high recall means that most actual defaulters are correctly identified. This approach minimizes the risk of financial loss due to unexpected defaults.**
**- Manage Precision: While maintaining a reasonable level of precision is important to avoid unnecessary interventions or lost opportunities, the primary focus should be on capturing as many potential defaulters as possible.**

**In summary, in the context of predicting loan default rates, recall is typically more critical than precision to protect the financial institution from significant risks associated with loan defaults.**

**The team will need to decide what the trade-offs will be in terms of going for a high accuracy or a higher recall**

```python
# Train the logistic regression model
model = LogisticRegression()
model.fit(X_train2, y_train)

# Predict probabilities on the test set
# The output will be an array with two columns: [probability of class 0, probability of class 1]
predicted_probabilities = model.predict_proba(X_test1)

# Get the probabilities for the positive class (default = 1)
prob_default = predicted_probabilities[:, 1]

# Print the predicted probabilities for the first 10 instances
print(prob_default[:10])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, f1_score, confusion_matrix

# Assuming you have the true labels and predicted probabilities
# y_test: actual default statuses (0 or 1)
# prob_default: predicted probabilities for the positive class (default = 1)

# Step 1: Calculate performance metrics for a range of thresholds
precision, recall, thresholds = precision_recall_curve(y_test, prob_default)

# Calculate F1 score for each threshold
f1_scores = 2 * (precision * recall) / (precision + recall)

# Step 2: Plot precision, recall, and F1 score against thresholds
plt.figure(figsize=(10, 6))
plt.plot(thresholds, precision[:-1], label='Precision')
plt.plot(thresholds, recall[:-1], label='Recall')
plt.plot(thresholds, f1_scores[:-1], label='F1 Score')
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.title('Precision, Recall, and F1 Score for Different Thresholds')
plt.legend()
plt.grid(True)
plt.show()

# Step 3: Choose the optimal threshold
optimal_idx = np.argmax(f1_scores)
optimal_threshold = thresholds[optimal_idx]

print(f'Optimal Threshold: {optimal_threshold}')
print(f'F1 Score at Optimal Threshold: {f1_scores[optimal_idx]}')

# Evaluate the model at the optimal threshold
y_pred_optimal = (prob_default >= optimal_threshold).astype(int)
conf_matrix = confusion_matrix(y_test, y_pred_optimal)

print('Confusion Matrix at Optimal Threshold:')
print(conf_matrix)
```
