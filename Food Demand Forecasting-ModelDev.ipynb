```python
#Seperating the target feature

X_train = train_df.drop(columns=['num_orders'])
y_train = train_df['num_orders']
```

```python
# Create rolling mean features for num_orders

# Sample data (assuming train_df and X_train are already defined)
# Ensure 'meal_id' is present in both DataFrames
if 'meal_id' in train_df.columns and 'meal_id' in X_train.columns:
    # Compute rolling means for num_orders
    rolling_mean_3 = train_df.groupby('meal_id')['num_orders'].rolling(3).mean().reset_index(level=0, drop=True).fillna(0)
    rolling_mean_7 = train_df.groupby('meal_id')['num_orders'].rolling(7).mean().reset_index(level=0, drop=True).fillna(0)

    # Add rolling mean features to X_train
    X_train['num_orders_rolling_mean_3'] = rolling_mean_3
    X_train['num_orders_rolling_mean_7'] = rolling_mean_7

    # Fill any remaining NaN values
    X_train['num_orders_rolling_mean_3'].fillna(0, inplace=True)
    X_train['num_orders_rolling_mean_7'].fillna(0, inplace=True)
else:
    print("Error: 'meal_id' column is not present in one of the DataFrames")
```

```python
# Create features for the difference between base price and checkout price
X_train['price_diff'] = X_train['base_price'] - X_train['checkout_price']

# Create features for the ratio of base price to checkout price
X_train['price_ratio'] = X_train['base_price'] / X_train['checkout_price']

# Create features for the interaction between emailer_for_promotion and homepage_featured
X_train['promotion_interaction'] = X_train['emailer_for_promotion'] + X_train['homepage_featured']

# One-hot encode categorical features
# Convert categorical features to numerical using one-hot encoding
X_train = pd.get_dummies(X_train, columns=['category', 'cuisine', 'city_code', 'region_code', 'center_type'])
```

```python
# Create lagged features for num_orders
X_train['num_orders_lag1'] = train_df.groupby(['meal_id'])['num_orders'].shift(1).fillna(0)
X_train['num_orders_lag2'] = train_df.groupby(['meal_id'])['num_orders'].shift(2).fillna(0)
```

```python
import datetime

# Create features for the day of the week and month
# Assuming 'week' is in the format 'YYYY-WW' (year and week number)
# If 'week' is an integer, convert it to string first and pad with zeros if necessary
X_train['week'] = X_train['week'].astype(str).apply(lambda x: x.zfill(2)) # Pad with zeros if week number is single digit
X_train['week'] = X_train['week'].apply(lambda x: '2018-' + x) # Assuming all data is from 2018, replace with actual year if different

# Handle weeks beyond 53 and potential multiple hyphens
def handle_week(week_str):
    parts = week_str.split('-')  # Split by hyphen
    if len(parts) == 2:  # Handle the standard case with two parts
        year, week_num = parts
        if int(week_num) > 53:
            year = str(int(year) + 1)
            week_num = '01'  # Start with week 1 of the next year
        return datetime.datetime.strptime(year + '-'+ week_num + '-1', "%Y-%W-%w")
    else:  # Handle cases with multiple hyphens
        # Implement logic to extract year and week number for these cases
        # For example, you might take the first and last parts as year and week number
        return datetime.datetime(2018, 1, 1)  # Return a default date instead of None

X_train['week'] = X_train['week'].apply(handle_week)

# Check if 'week' column contains datetime objects
print(X_train['week'].dtype)

# If not datetime, convert to datetime
if X_train['week'].dtype != 'datetime64[ns]':
    X_train['week'] = pd.to_datetime(X_train['week'])

X_train['day_of_week'] = X_train['week'].dt.dayofweek
X_train['month'] = X_train['week'].dt.month

# Create features for the number of orders per center per week
center_week_count = X_train.groupby(["center_id", "week"])["id"].count().reset_index()
center_week_count.columns = ["center_id", "week", "center_week_count"]
X_train = pd.merge(X_train, center_week_count, on=["center_id", "week"], how="left")

# Find the one-hot encoded category columns
category_columns = [col for col in X_train.columns if col.startswith('category_')]

# Iterate over one-hot encoded category columns and create features
for category_col in category_columns:
    category_week_count = X_train.groupby([category_col, "week"])["id"].count().reset_index()
    category_week_count.columns = [category_col, "week", category_col + "_week_count"]
    X_train = pd.merge(X_train, category_week_count, on=[category_col, "week"], how="left")

# Create features for the average number of orders per meal in the last 4, 10 and 20 weeks
X_train['meal_avg_orders_4w'] = train_df.groupby('meal_id')['num_orders'].transform(lambda x: x.rolling(4).mean().shift(1))
X_train['meal_avg_orders_10w'] = train_df.groupby('meal_id')['num_orders'].transform(lambda x: x.rolling(10).mean().shift(1))
X_train['meal_avg_orders_20w'] = train_df.groupby('meal_id')['num_orders'].transform(lambda x: x.rolling(20).mean().shift(1))

# Create features for the average price difference per meal in the last 4, 10 and 20 weeks
X_train['meal_avg_price_diff_4w'] = X_train.groupby('meal_id')['price_diff'].transform(lambda x: x.rolling(4).mean().shift(1))
X_train['meal_avg_price_diff_10w'] = X_train.groupby('meal_id')['price_diff'].transform(lambda x: x.rolling(10).mean().shift(1))
X_train['meal_avg_price_diff_20w'] = X_train.groupby('meal_id')['price_diff'].transform(lambda x: x.rolling(20).mean().shift(1))

# Fill missing values in newly created features
X_train.fillna(0, inplace=True)
```

```python
X = X_train.copy()
y = y_train.copy()
```

**Splitting**

```python
#splitting the data first into two part -- doing a 70:30 split i.e 30% data fed to intermediate test data set

from sklearn.model_selection import train_test_split
x_train1,x_test,y_train1,y_test = train_test_split(X,y,test_size=0.3,random_state=0 )
```

```python
x_train1.shape ,x_test.shape ,y_train1.shape , y_test.shape
```

```python
# **XGBoost**

import xgboost as xgb
import pandas as pd

# Convert 'week' column to numerical representation (e.g., ordinal)
# Assuming 'week' represents a week number within a year
x_train1['week'] = x_train1['week'].dt.isocalendar().week
x_test['week'] = x_test['week'].dt.isocalendar().week

# Create XGBoost regressor object
xgb_model = xgb.XGBRegressor(
    n_estimators=100,  # Number of boosting rounds
    learning_rate=0.1,  # Step size shrinkage used in update to prevent overfitting
    max_depth=10,  # Maximum depth of a tree
    subsample=0.8,  # Subsample ratio of the training instances
    colsample_bytree=0.8,  # Subsample ratio of columns when constructing each tree
    random_state=42  # Random number seed for reproducibility
)

# Fit the model to the training data
xgb_model.fit(x_train1, y_train1)

# Make predictions on the test data
xgb_preds = xgb_model.predict(x_test)

# **Evaluation**
from sklearn.metrics import mean_squared_error

# Calculate RMSE for XGBoost
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_preds)) # Fixed variable name here from xgb_predictions to xgb_preds
print("XGBoost RMSE:", xgb_rmse)

import numpy as np
from sklearn.metrics import mean_squared_log_error

def rmsle(y_true, y_pred):
  """
  Calculates the Root Mean Squared Logarithmic Error (RMSLE)

  Args:
    y_true: Array-like, true target values.
    y_pred: Array-like, predicted target values.

  Returns:
    The RMSLE score.
  """
  y_true_clipped = np.clip(y_true, a_min=0, a_max=None)
  y_pred_clipped = np.clip(y_pred, a_min=0, a_max=None)
  return np.sqrt(mean_squared_log_error(y_true_clipped, y_pred_clipped))

# Example usage (assuming y_test and xgb_preds are defined):
rmsle_score = rmsle(y_test, xgb_preds)
print("RMSLE:", rmsle_score)

from sklearn.metrics import mean_squared_log_error
xgb_rmsle = rmsle(y_test, xgb_preds)
print("XGBoost RMSLE:", xgb_rmsle)
```

```python
import lightgbm as lgb
import numpy as np

def rmsle(preds, data):
    """ Root Mean Squared Logarithmic Error """
    labels = data.get_label()
    preds = np.maximum(0, preds)  # Ensure there are no negative values
    rmsle_value = np.sqrt(np.mean((np.log1p(preds) - np.log1p(labels))**2))
    return 'rmsle', rmsle_value, False

# Assuming X_train, y_train, X_test, y_test are already defined

# Create LightGBM datasets
train_data = lgb.Dataset(x_train1, label=y_train1)
valid_data = lgb.Dataset(x_test, label=y_test, reference=train_data)

# Define the parameters
params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'learning_rate': 0.05,
    'num_leaves': 31,
    'verbose': -1
}

# Train the LightGBM model with early stopping and custom metric
lgb_model = lgb.train(
    params,
    train_data,
    num_boost_round=1000,
    valid_sets=[train_data, valid_data],
    valid_names=['train', 'valid'],
    feval=rmsle,
    callbacks=[lgb.early_stopping(stopping_rounds=10)]
)

# Predict using the trained model
y_pred = lgb_model.predict(x_test, num_iteration=lgb_model.best_iteration)
```
